```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import Model, Input
from tensorflow.keras.layers import Dense, BatchNormalization, Dropout, Concatenate, Layer
from tensorflow.keras.optimizers import Adam
from sklearn.preprocessing import StandardScaler

# ----------------------
# Step 1: Normalize Inputs
# ----------------------
def normalize_data(X):
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    return pd.DataFrame(X_scaled, index=X.index, columns=X.columns), scaler

X_expr, scaler_expr = normalize_data(X_expr)
X_cnv, scaler_cnv = normalize_data(X_cnv)
X_meth, scaler_meth = normalize_data(X_meth)

# ----------------------
# Step 2: Custom Sampling Layer with KL Loss
# ----------------------
class Sampling(Layer):
    def call(self, inputs):
        z_mean, z_log_var = inputs
        epsilon = tf.random.normal(shape=tf.shape(z_mean))
        z = z_mean + tf.exp(0.5 * z_log_var) * epsilon

        # Compute KL divergence loss and add it to model losses
        kl_loss = -0.5 * tf.reduce_mean(
            tf.reduce_sum(1 + z_log_var - tf.square(z_mean) - tf.exp(z_log_var), axis=1)
        )
        self.add_loss(kl_loss)
        return z

# ----------------------
# Step 3: Encoder and Decoder Builders
# ----------------------
def build_encoder(input_shape, latent_dim, name_suffix):
    inputs = Input(shape=(input_shape,))
    x = Dense(128, activation='relu')(inputs)
    x = BatchNormalization()(x)
    x = Dropout(0.2)(x)
    z_mean = Dense(latent_dim, name=f"z_mean_{name_suffix}")(x)
    z_log_var = Dense(latent_dim, name=f"z_log_var_{name_suffix}")(x)
    z = Sampling(name=f"z_{name_suffix}")([z_mean, z_log_var])
    return Model(inputs, [z_mean, z_log_var, z], name=f"encoder_{name_suffix}")

def build_decoder(latent_dim, output_shape):
    latent_inputs = Input(shape=(latent_dim,))
    x = Dense(128, activation='relu')(latent_inputs)
    x = BatchNormalization()(x)
    x = Dropout(0.2)(x)
    outputs = Dense(output_shape, activation='linear')(x)
    return Model(latent_inputs, outputs, name="decoder")

# ----------------------
# Step 4: Build VAE Model
# ----------------------
latent_dim = 32

input_expr = Input(shape=(X_expr.shape[1],), name="input_expr")
input_cnv = Input(shape=(X_cnv.shape[1],), name="input_cnv")
input_meth = Input(shape=(X_meth.shape[1],), name="input_meth")

encoder_expr = build_encoder(X_expr.shape[1], latent_dim, "expr")
encoder_cnv = build_encoder(X_cnv.shape[1], latent_dim, "cnv")
encoder_meth = build_encoder(X_meth.shape[1], latent_dim, "meth")

z_mean_expr, z_log_var_expr, z_expr = encoder_expr(input_expr)
z_mean_cnv, z_log_var_cnv, z_cnv = encoder_cnv(input_cnv)
z_mean_meth, z_log_var_meth, z_meth = encoder_meth(input_meth)

# Concatenate latent vectors from all modalities
z_fused = Concatenate(name="z_fused")([z_expr, z_cnv, z_meth])

# Decoder for the concatenated latent space
decoder = build_decoder(latent_dim * 3, X_expr.shape[1] + X_cnv.shape[1] + X_meth.shape[1])
recon = decoder(z_fused)

vae = Model(inputs=[input_expr, input_cnv, input_meth], outputs=recon, name="multimodal_vae")

# ----------------------
# Step 5: Loss Function (Reconstruction only, KL handled in Sampling)
# ----------------------
def vae_loss(x_true, x_pred):
    return tf.reduce_mean(tf.square(x_true - x_pred))

# Build and compile your VAE as before
vae.compile(optimizer=Adam(1e-4), loss=vae_loss)

# Prepare training inputs/targets
X_input = np.concatenate([X_expr.values, X_cnv.values, X_meth.values], axis=1)

# Train the model
vae.fit(
    [X_expr.values, X_cnv.values, X_meth.values],
    X_input,
    epochs=150,
    batch_size=16,
    shuffle=True
)

# Save weights after training
vae.save_weights("vae_multimodal.weights.h5")
encoder_expr.save_weights("encoder_expr.weights.h5")
encoder_cnv.save_weights("encoder_cnv.weights.h5")
encoder_meth.save_weights("encoder_meth.weights.h5")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# =======================
# Load datasets
# =======================
df_hiseqv2 = pd.read_csv("/content/drive/MyDrive/RI/hiseqv2_common_samples.csv")
df_meth = pd.read_csv("/content/drive/MyDrive/RI/meth_common_samples.csv")
df_gistic = pd.read_csv("/content/drive/MyDrive/RI/gistic_common_samples.csv")
df_clinical = pd.read_csv("/content/drive/MyDrive/RI/clinical_common_samples.csv")

# =======================
# Map survival status
# =======================
surv_map = {0: "Alive", 1: "Dead"}
df_clinical["vital_status"] = df_clinical["vital_status"].map(surv_map)

# =======================
# Function to plot heatmap
# =======================
def plot_survival_heatmap(omics_df, clinical_df, title):
    gene_col = omics_df.columns[0]

    # Align samples
    common_samples = list(set(omics_df.columns[1:]) & set(clinical_df["sampleID"]))
    if len(common_samples) == 0:
        print(f"No common samples found for {title}. Skipping.")
        return

    omics_df = omics_df[[gene_col] + common_samples]

    # Align and filter clinical data
    clinical_df = clinical_df.set_index("sampleID").loc[common_samples]
    clinical_df = clinical_df.dropna(subset=["vital_status"])
    if clinical_df.shape[0] == 0:
        print(f"No samples with valid vital_status for {title}. Skipping.")
        return

    omics_df = omics_df[[gene_col] + clinical_df.index.tolist()]

    # Prepare matrix for heatmap
    data_matrix = omics_df.drop(columns=[gene_col]).astype(float)
    if data_matrix.shape[1] == 0:
        print(f"No data to plot for {title}. Skipping.")
        return

    # Plot heatmap
    plt.figure(figsize=(12, 6))
    sns.heatmap(data_matrix, cmap="coolwarm", yticklabels=False, cbar_kws={'label': 'Expression'})
    plt.title(f"{title} - Survival Status")
    plt.xlabel("Samples")
    plt.ylabel("Genes/Probes")

    # Add survival status color bars
    status_colors = clinical_df["vital_status"].map({"Alive": "green", "Dead": "red"})
    for idx, color in enumerate(status_colors):
        plt.gca().add_patch(plt.Rectangle((idx, 0), 1, data_matrix.shape[0],
                                          fill=False, edgecolor=color, lw=1))
    plt.show()

# =======================
# Plot for each omics type
# =======================
plot_survival_heatmap(df_meth, df_clinical, "Methylation")
plot_survival_heatmap(df_hiseqv2, df_clinical, "HiSeq RNA-seq")
plot_survival_heatmap(df_gistic, df_clinical, "CNV")
```

```python
import requests
import networkx as nx
import matplotlib.pyplot as plt

# Get top 50 genes from your ranked DataFrame
top_genes = df_ranked['Gene'].head(50).tolist()

# Join genes with %0d as required by STRING API for multiple identifiers
identifiers = "%0d".join(top_genes)

# STRING API URL for network query with a moderate confidence threshold (200)
url = (
    f"https://string-db.org/api/tsv/network"
    f"?identifiers={identifiers}&species=9606&required_score=200"
)

# Request interaction data from STRING
response = requests.get(url)
if response.status_code != 200 or not response.text.strip():
    raise Exception("STRING API request failed or returned empty.")

# Parse the TSV response
lines = response.text.strip().split("\n")
header = lines[0].split("\t")
edges = [line.split("\t") for line in lines[1:]]

# Build graph including only edges between your top genes
G = nx.Graph()
for edge in edges:
    gene1 = edge[2].strip()  # preferredName_A
    gene2 = edge[3].strip()  # preferredName_B
    score = float(edge[5])   # combined_score

    if gene1 in top_genes and gene2 in top_genes:
        G.add_edge(gene1, gene2, weight=score)

print(f"STRING network has {G.number_of_nodes()} nodes and {G.number_of_edges()} edges")

if G.number_of_nodes() > 0:
    plt.figure(figsize=(14, 12))
    pos = nx.spring_layout(G, k=0.3, seed=42)

    # Node size scaled by degree (number of connections)
    node_sizes = [100 + 300 * G.degree(n) for n in G.nodes()]
    nx.draw_networkx_nodes(G, pos, node_size=node_sizes, node_color='skyblue', alpha=0.8)
    nx.draw_networkx_labels(G, pos, font_size=9)

    # Edge width scaled by interaction confidence score
    edge_weights = [d['weight'] * 5 for (u, v, d) in G.edges(data=True)]
    nx.draw_networkx_edges(G, pos, width=edge_weights, alpha=0.6)

    plt.title("STRING Network of Top 50 Genes by SHAP Importance")
    plt.axis('off')
    plt.show()
else:
    print("No interactions found with the current score threshold.")
```

```python
import numpy as np
import pandas as pd
from scipy.stats import ttest_ind

# Inputs:
# X_expr: DataFrame (samples x genes) with expression data
# cluster_labels: array-like of cluster assignments per sample

# 1. Filter low variance genes
gene_vars = X_expr.var(axis=0)
high_var_genes = gene_vars[gene_vars > 0.1].index
X_expr_filtered = X_expr[high_var_genes]

# 2. Prepare DE results container
de_results = []

# Unique clusters
clusters = np.unique(cluster_labels)

for cluster in clusters:
    group1_idx = np.where(cluster_labels == cluster)[0]
    group2_idx = np.where(cluster_labels != cluster)[0]

    expr_group1 = X_expr_filtered.iloc[group1_idx]
    expr_group2 = X_expr_filtered.iloc[group2_idx]

    # Perform gene-wise t-test
    pvals = []
    fold_changes = []
    for gene in X_expr_filtered.columns:
        stat, p = ttest_ind(expr_group1[gene], expr_group2[gene], equal_var=False)
        pvals.append(p)
        # Fold change = mean(cluster) - mean(other)
        fc = expr_group1[gene].mean() - expr_group2[gene].mean()
        fold_changes.append(fc)

    # Multiple testing correction (Benjamini-Hochberg)
    pvals = np.array(pvals)
    sorted_indices = np.argsort(pvals)
    sorted_pvals = pvals[sorted_indices]
    n = len(pvals)
    bh_adjusted = np.empty(n, dtype=float)
    prev_bh_value = 0
    for i in reversed(range(n)):
        rank = i + 1
        bh_value = sorted_pvals[i] * n / rank
        bh_value = min(bh_value, prev_bh_value if prev_bh_value > 0 else 1)
        bh_adjusted[i] = bh_value
        prev_bh_value = bh_value
    adj_pvals = np.empty(n, dtype=float)
    adj_pvals[sorted_indices] = bh_adjusted

    # Compile results for this cluster
    for i, gene in enumerate(X_expr_filtered.columns):
        de_results.append({
            'gene': gene,
            'pval': pvals[i],
            'adj_pval': adj_pvals[i],
            'fold_change': fold_changes[i],
            'cluster': cluster
        })

de_results_df = pd.DataFrame(de_results)

# 3. Filter significant DE genes with relaxed threshold (adj_pval < 0.1, |fc| > 0.3)
sig_de_genes = de_results_df[(de_results_df['adj_pval'] < 0.1) & (np.abs(de_results_df['fold_change']) > 0.3)]

print(f"Significant DE genes with relaxed thresholds: {len(sig_de_genes)}")

# Optional: display top 10 by adjusted p-value
print(sig_de_genes.sort_values('adj_pval').head(10))
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

event_col = 'new_tumor_event_after_initial_treatment'  # survival/event status

# Automatically detect numeric columns per data type
def detect_numeric_cols(df, keywords):
    cols = [col for col in df.columns if any(k in col for k in keywords)]
    numeric_cols = df[cols].apply(pd.to_numeric, errors='coerce').dropna(axis=1, how='all').columns.tolist()
    return numeric_cols

# General heatmap plotting function
def plot_survival_heatmap_auto(df, keywords, event_col, title):
    data_cols = detect_numeric_cols(df, keywords)
    if not data_cols:
        print(f"[SKIP] No usable numeric columns found for {title}")
        return

    df_numeric = df[data_cols]

    # Encode survival status
    df['Survival_Status'] = df[event_col].apply(
        lambda x: 'Dead' if str(x).upper() == 'YES' or x == 1 else 'Survived'
    )

    # Prepare data for plotting
    df_plot = pd.concat([df_numeric, df['Survival_Status']], axis=1)
    df_plot_sorted = df_plot.sort_values(by='Survival_Status')
    expr_matrix = df_plot_sorted[data_cols].T
    status_colors = df_plot_sorted['Survival_Status'].map({'Survived': 'green', 'Dead': 'red'})

    sns.clustermap(expr_matrix, cmap='RdBu_r', col_colors=status_colors,
                   xticklabels=False, yticklabels=True,
                   row_cluster=False, col_cluster=False,
                   figsize=(14, 8),
                   cbar_kws={'label': f'{title} Level'})
    plt.title(f"{title} Heatmap by Survival Status")
    plt.show()

# Define keywords for each omics type
meth_keywords = ['methyl', 'DNAMethyl', 'hmethyl']
hiseq_keywords = ['HiSeq', 'RNA', 'exp']
cnv_keywords = ['CNV', 'copy_number', 'PDMarray']

# Plot heatmaps
plot_survival_heatmap_auto(df_merged, meth_keywords, event_col, "Methylation")
plot_survival_heatmap_auto(df_merged, hiseq_keywords, event_col, "HiSeq Expression")
plot_survival_heatmap_auto(df_merged, cnv_keywords, event_col, "Copy Number Variation")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

def plot_survival_heatmap(omics_df, clinical_df, id_col, event_col, time_col, prefix_list, title):
    # Keep only columns with the given prefixes
    gene_cols = [col for col in omics_df.columns if any(col.startswith(p) for p in prefix_list)]
    if not gene_cols:
        print(f"[SKIP] No columns found for {title}")
        return

    # Merge on patient/sample ID
    merged_df = pd.merge(clinical_df, omics_df[gene_cols + [id_col]], on=id_col, how="inner")

    # Remove NaNs
    merged_df = merged_df.dropna(subset=gene_cols + [event_col, time_col])
    if merged_df.empty:
        print(f"[SKIP] No non-NaN data for {title}")
        return

    # Classify survival
    merged_df["SurvivalStatus"] = merged_df[event_col].apply(lambda x: "Dead" if x == 1 else "Alive")

    # Mean expression per survival status
    heatmap_data = merged_df.groupby("SurvivalStatus")[gene_cols].mean()

    plt.figure(figsize=(12, 6))
    sns.heatmap(heatmap_data, cmap="coolwarm", annot=False)
    plt.title(f"{title} - Survival Heatmap")
    plt.xlabel("Genes")
    plt.ylabel("Survival Status")
    plt.tight_layout()
    plt.show()

# Example usage
id_col = "bcr_patient_barcode"  # change if your ID column is different
event_col = "vital_status"      # 1=dead, 0=alive (adjust if needed)
time_col = "days_to_death"      # or days_to_last_followup

# Prefixes for your datasets
meth_prefixes = ["_GENOMIC_ID_TCGA_COAD_hMethyl27"]
hiseq_prefixes = ["_GENOMIC_ID_TCGA_COAD_HiSeq"]
cnv_prefixes = ["_GENOMIC_ID_TCGA_COAD_CNV"]

plot_survival_heatmap(df_meth_filtered , df_clinical_filtered, id_col, event_col, time_col, meth_prefixes, "Methylation")
plot_survival_heatmap(df_hiseqv2_filtered ,  df_clinical_filtered, id_col, event_col, time_col, hiseq_prefixes, "HiSeq RNA")
plot_survival_heatmap(df_gistic_filtered ,  df_clinical_filtered, id_col, event_col, time_col, cnv_prefixes, "CNV")
```
